````python
import csv
from IPython.display import display, Latex, Markdown

latex_cols = 4
fields_per_cell = 4

def note_value(note):
    return int(note[0]) + 7 * note.count("°")

def hand(note):
    ## 0: left hand, 1: right hand
    return note_value(note) % 2

## Format notes played by the left right hand
# def left_hand(x):
#         return rf"{x}"
# def right_hand(x):
#         return rf"{x}"

def format_cell(value):
    # digit_count = sum(c.isdigit() for c in value)
    notes = re.findall(r'[1-7]°*', value)
    hands = "".join([">" if hand(note) else "<" for note in notes])
    ## 1(23)4(567°) count as 4 events
    events = re.findall(r"\([^)]*\)|\d", value)
    event_count = len(events)
    ## 𝅘𝅥 quarter note
    if event_count == 1:
        value = value.replace("[", r"{\bf\LARGE{").replace("]", r"}}")
    ## 𝅘𝅥𝅮 eighth note
    elif event_count == 2:
        value = r"\underline{" + value + "}"
        value = value.replace("[", r"{\bf\large{").replace("]", r"}}")
    ## 𝅘𝅥𝅯 sixteenth note
    elif event_count == 4:
        value = r"\underline{\underline{" + value + "}}"
        value = value.replace("[", r"{\bf").replace("]", r"}")
    ## upper octave dot
    value = re.sub(r"(.)°", r"\\dot{\1}", value)
    return rf"\underset{{\scriptstyle{hands}}}{{{value}}}"

def make_cell(values):
    ## E.g. \begin{array}{llll}{1 & 2 & 3 & 4}\end{array}
    cell = r"\begin{array}{" + r"".join([r"l" for _ in range(fields_per_cell)]) + r"}"
    cell += " & \n".join(format_cell(value if value else "-") for value in values)
    cell += r"\end{array}"
    return cell

def generate_latex_row(csv_file):
    out = []
    ## E.g. \begin{array}{|l|l|l|l|}
    out.append(r"\begin{array}{|" + r"|".join([r"l" for _ in range(latex_cols)])+ r"|}")
    out.append(r"\hline")
    with open(csv_file, newline="", encoding="utf-8") as f:
        f_lines = csv.reader(f)
        cells = []
        for i, f_line in enumerate(f_lines, start=1):
            out.append(r"\rule{0pt}{4ex}")
            cell = make_cell(f_line)
            cells.append(cell)
            if i%latex_cols == 0:
                out.append(" & \n".join(cells))
                out.append(r"\\")
                out.append(r"\hline")
                cells = []
    out.append(r"\end{array}")
    return "\n".join(out)

latex = generate_latex_row("input.csv")
display(Markdown(markdown))
display(Latex(r"$" + latex + r"$"))
display(Markdown(markdown_foot))
display(Markdown(r"<br><br><br>"))
display(Latex(r"$" + latex_melody + r"$"))
````

````python
from IPython.display import display, Latex
# 左右手标记
def left_hand(x):
    return rf"{x}_{{\tiny L}}"
def right_hand(x):
    return rf"{x}_{{\tiny R}}"
data = r"\dot3 1 3 5"
notes = data.split()
result = []
for i, note in enumerate(notes):
    if i == 0:
        result.append(left_hand(note))
    else:
        result.append(right_hand(note))
latex = r"\quad ".join(result)
display(Latex(latex))
````

````python
from IPython.display import display, Latex
## 左右手标记
def left_hand(x, is_underset=False):
    if is_underset:
        return rf"\llap<{x}"
    else:
        return rf"\llap{{\scriptstyle<}}{x}"
def right_hand(x, is_underset=False):
    if is_underset:
        return rf"{x}\rlap{{>}}"
    else:
        return rf"{x}\rlap{{\scriptstyle>}}"
l, r = left_hand(r"\bf\LARGE\dot3"), right_hand(r"(135)", True)
latex = rf"\quad\underset{{{r}}}{{{l}}}"
l, r = left_hand(r"2", True), right_hand(r"\bf(57{\LARGE\dot2})")
latex += rf"\quad\underset{{{l}}}{{{r}}}"
display(Latex(latex))
````
