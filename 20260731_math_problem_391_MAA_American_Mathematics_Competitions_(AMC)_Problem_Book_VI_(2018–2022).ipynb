# **Solution**

* Let the side length of $S_n$ be $s_n$. Since $s_{n+1}=\frac12s_n$ and the horizontal diagonal of $S_2$ is $3$, $s_2=\frac{3}{\sqrt2}$, $s_1=2s_2=3\sqrt2$.

* Hence the area of $S_1$ is $A_1=(3\sqrt2)^2=18$.  

* The remaining squares have areas forming a geometric series with ratio $\frac14$:   
$A_2+A_3+\cdots
=18\left(\frac14+\frac1{16}+\cdots\right)
=18\cdot\frac{\frac14}{1-\frac14}
=6$.

* Therefore, the total area is $18+6=24$, so each side of the dividing line must contain area $12$.

* Since every square after $S_1$ lies entirely to the right of $x=6$, the line $x=t$ (with $t<6$) cuts only $S_1$. Thus the portion of $S_1$ to the left of the line must have area $12$.

* The horizontal diagonal of $S_1$ is $6$, so its vertices are $(0,0),\ (3,3),\ (6,0),\ (3,-3)$.  

* The line $x=t$ removes a right isosceles triangle whose horizontal width is $6-t$ and vertical height is $2(6-t)$. Its area is $\frac12\bigl(2(6-t)\bigr)(6-t)=(6-t)^2$.

* Hence, $18-(6-t)^2=12$, so $(6-t)^2=6$, $t=6-\sqrt6$.  

# 👉 **Answer**

* **Choice C**: $\boxed{6-\sqrt6}$
